# Magnifica Humanitas: Key Points

*Magnifica Humanitas* is an encyclical letter issued by Pope Leo XIV on 15 May 2026, addressing how the Church's Social Doctrine should guide humanity through the age of artificial intelligence. Building on the tradition begun by Leo XIII's *Rerum Novarum* in 1891, Leo XIV frames the choice before us using two biblical images: the Tower of Babel, built on pride and self-sufficiency, versus the rebuilding of Jerusalem under Nehemiah, built on shared responsibility and communion. The encyclical argues that the real question isn't whether to use AI, but what kind of world we're building with it — one that safeguards human dignity, work, truth, and peace, or one that lets efficiency and power dominate the human person.

## Framing: Two Biblical Images (Introduction)

The whole encyclical runs on one governing metaphor:

- **Tower of Babel** (Gen 11): a "single language, single technology, single direction" built on pride and self-sufficiency — efficiency without God, ending in dehumanization and dispersion.
- **Rebuilding Jerusalem under Nehemiah** (Neh 2–6): shared responsibility, listening, everyone assigned "their own section of the wall," God at the center — communion, not uniformity.

Leo's core claim: the choice isn't "yes or no to technology" but **Babel or Jerusalem** — domination vs. shared, humble rebuilding. This is a great discussion-starter for CIAI since it reframes "is AI good or bad" as the wrong question.



## Chapter 1: A Church That Journeys Through History

- Social Doctrine isn't a fixed rulebook — it's a *living tradition* that develops as each pope responds to the "new things" (*res novae*) of his era, starting with Leo XIII's *Rerum Novarum* (1891).
- Quick tour of predecessor encyclicals (useful as a reference chart for students): Rerum Novarum → dignity of work; Quadragesimo Anno → subsidiarity; Pacem in Terris → human rights framework; Gaudium et Spes → Church's dialogue with the modern world; Populorum Progressio → integral development; Laudato Si' → integral ecology/technocratic paradigm; Fratelli Tutti → social friendship.
- Key move: AI isn't just "another topic" for CST to comment on — it *challenges the categories of Social Doctrine from within*.



## Chapter 2: Foundations and Principles

This chapter is your **core CST vocabulary chapter** — probably the most directly reusable for your course.

**Foundation:** Every person is made in the image of the Triune God, possessing **infinite, ontological dignity** — not earned, not dependent on productivity or capability (¶50–53). This directly counters any AI-era logic that ranks human worth by output or efficiency.

**Five Principles** (each explicitly applied to the digital/AI context):
1. **Common Good** — society is "greater than the sum of its parts"; not a pile of individual interests.
2. **Universal Destination of Goods** — extended explicitly to *new* goods: "patents, algorithms, digital platforms, technological infrastructure and data" (¶67) — concentration of these in a few hands violates the principle.
3. **Subsidiarity** — reframed for the digital age: the "highest level" of power today isn't the State but **tech companies/platforms** who set the rules of access and visibility; subsidiarity demands transparency, accountability, and genuine participation from communities (¶71–72).
4. **Solidarity** — includes recognizing the "hidden, often exploited workers who sustain algorithmic systems" (data labelers, content moderators).
5. **Social Justice** — includes fighting algorithmic bias, surveillance, and unequal access to technology; preferential option for the poor extended to the "digitally excluded."



## Chapter 3: Technology and Dominance — The Grandeur of Humanity

The philosophical/theological heart of the document.

- **The technocratic paradigm** (borrowed from *Laudato Si'*): letting efficiency, control, and profit alone drive decisions, reducing persons to "cogs."
- **What AI actually is** (¶97–99) — very quotable for class: AI systems are "cultivated" rather than "built"; even developers don't fully understand their internal workings. AI **imitates** but does not *possess* intelligence in the human sense — no body, no experience, no moral conscience, no capacity for love, forgiveness, or responsibility. It can *simulate* empathy but not *have* it.
- **Risks of personal use** (¶100): over-reliance, false sense of objectivity, and the danger of *simulated relationship* replacing real human connection.
- **AI is never morally neutral** (¶104) — every system embeds choices about what it measures, ignores, and optimizes.
- **"Disarming AI"** (¶110) — a key original phrase: freeing AI from an "armed competition" mentality (military, economic, cognitive), opening it to plurality rather than monopoly.
- **Transhumanism/posthumanism critique** (¶115–117): the danger isn't technology itself but the *underlying vision* — if the human is something to be "surpassed," it becomes easier to treat some lives as less worthy.
- **The value of limits** (¶118–122): suffering, finitude, and vulnerability aren't defects to eliminate — they're where compassion, love, and meaning actually grow. "To eliminate suffering entirely would mean extinguishing love and desire as well."
- **The authentic "more than human"** is Christian grace, not technological enhancement (¶127–128): "We become fully human when we become more than human" through God's grace — not self-sufficiency.
- Closes with Augustine's **two loves, two cities** — self-love vs. love of God/neighbor — as the real choice underlying every tech decision.



## Chapter 4: Truth, Work, Freedom

Practical/social applications — good for case-study discussions:

- **Truth as a common good**: AI-amplified disinformation threatens democracy itself; truth is relational and verified, not centrally controlled.
- **Education**: calls for a "renewed educational alliance" between families, schools, and Church; explicit warnings about early/unsupervised device use, online exploitation risks for minors, and AI's threat to the "desire to ask questions."
- **Work**: automation should serve human dignity, not just efficiency; work is a "normal path toward maturity," not just income. Warns against AI systems that force workers to adapt to machine pace rather than vice versa.
- **New forms of slavery**: explicit, strong language on exploited data-labeling/content-moderation workers, rare-earth mineral extraction (including child labor), human trafficking via digital platforms, and "data colonialism" (health/genetic data extraction from vulnerable regions).
- **Notable moment**: Leo XIV formally apologizes for the Church's historical slowness in condemning slavery (¶176) as a caution against repeating blindness in the AI era.



## Chapter 5: The Culture of Power vs. The Civilization of Love

- AI in warfare: explicitly forbids delegating **lethal decisions to autonomous systems** — "no algorithm can make war morally acceptable" (¶198). Lays out concrete criteria: identifiable responsibility, protection of civilians, human control must remain "effective, self-aware, and responsible."
- Diagnoses a "culture of power": normalization of war, arms races (including AI weapons), crisis of multilateral institutions, and "Realpolitik" as false realism.
- Offers **five practical paths** for ordinary people (great for a student reflection exercise): disarm your words, pursue justice, adopt victims' perspective, cultivate healthy realism, revive dialogue.


## Conclusion

Four-part "program for Christian life" in the AI era: contemplate God's plan, live Eucharistic unity, build for the common good, pray with Mary (the Magnificat as a lens for seeing history from "below" rather than from power).

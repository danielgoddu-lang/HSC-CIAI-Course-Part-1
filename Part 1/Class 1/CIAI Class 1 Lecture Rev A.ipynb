## 2. Define Artificial Intelligence (AI)

What is AI to you? Define AI in your own words.


When scientists speak of artificial intelligence, they are referring to the ways that a computer can be `intelligent` like humans - literally.  It has little or nothing to do with human intelligence. Instead, scientists ask: Can a computer display `intelligence`?  How would a computer demonstrate that it is `intelligent`? What would it have to do?

How do humans demonstrate `intelligence`? 

- **Learning:** Obtain, process, and maintain new information
- **Reasoning:** Manipulate and judge information
- **Understanding:** Analyze results of new information
- **Grasp truth:** Validate processed information
- **See relationships:** How data interacts with other data
- **Consider meaning:** Apply truths to situations
- **Separate facts from belief:** Determine if the data is adequately and consistently supports conclusions

AI Scientists categorize AI in four ways:

- Acting humanly

    When a user can't differentiate between a computer and a human (Turing test)

- Thinking humanly

    When a computer performs a task like a human (self-driving cars)
- Thinking rationally

    When a computer solves problems logically like a human (checkers, chess)

- Acting rationally

    When a computer interacts with its environment to make a decision in real-time (logically deviating from normal processing when required)

This all leads to the main and crucial part of AI:

> ## AI is about building `agents` that **perceive** and **act** to achieve goals. Agents perceive its environment through `sensors` and act using `actuators`.

Examples:

- thermostat
- self-driving car
- ChatGPT

### 2A: AI Capabilities

* `Narrow` AI that is good at one specific task (or a narrow set of tasks)

    * Examples are: spam filters, fraud detection models, recommendation systems, voice assistants, image classifiers, generative AI (ChatGPT, Copilot, Claude, etc.)
    * Each of these systems can only perform well in its own domain
    * Most AI systems today are `narrow` or `weak` AI
* `General` AI can learn and performan any intellectual taks a human can
    * Not a pre-programmed task
    * Examples are: learn a new job without training, reason across domains, transfer knowledge
* `Superintelligence` AI that is more capable that humans at essentially all intellectual tasks
    * Can reason, create, discover, strategize, and socially understand all on its own
    * Far beyond humans


## 3. Distinguish between Computer Science, AI, Machine Learning (ML), and Deep Learning (DL).

AI is the broad concept of machines doing intelligent things.

- Machine Learning (ML) is a subset of AI that uses data and algorithms to "learn" from experience using `labels`.

- Deep Learning (DL) is a further subset of ML that uses artificial neural networks to model complex patterns.
- Analogy: Think of AI as the universe, ML as a galaxy, and DL as a solar system within that galaxy.
- Supervised Learning is a type of machine learning where a model is trained using **labeled data** — that is, data where the correct answers are already known. In this approach, the model learns by comparing its predictions to the correct answers and gradually improving over time. For example, if you train a model with many emails labeled “spam” or “not spam,” it learns patterns that help it classify new emails correctly. *You show the computer examples with the right answers, and it learns to predict those answers on its own.*

- Unsupervised Learning is a type of machine learning where a model is given **unlabeled data** and must find patterns, groupings, or structure on its own. Unlike supervised learning, no one tells the model the “right answer” ahead of time. Instead, it looks for similarities or hidden relationships in the data. For example, a model might group customers into different categories based on buying behavior without being told what those categories should be. *The computer is given a pile of data and asked to discover what patterns are there by itself.*

Inside artificial intelligence is **Machine Learning**, which is a more specific area where computers learn patterns from data instead of being told every rule directly by a programmer. For example, instead of programming every rule for identifying spam email, a machine learning system studies many examples of spam and non-spam messages and learns the patterns for itself.

Inside machine learning is **Deep Learning**, which is an even more specialized approach that uses layers of artificial “neurons” called neural networks. Deep learning is what powers many of today’s most advanced AI systems, such as image recognition, voice assistants, and large language models like ChatGPT.

The **Data Science** circle overlaps with AI and machine learning because these fields often use the same tools and data. However, data science is not exactly the same as AI. Data science is more focused on collecting, organizing, analyzing, and drawing conclusions from data. A data scientist might use machine learning, but they might also just use statistics, charts, and spreadsheets to answer questions.

The main idea students should understand is this: **AI is part of computer science, machine learning is part of AI, and deep learning is part of machine learning.** Data science overlaps with all of them because data is essential to how modern AI works.

## 6. Catholic Teaching on AI: Pope Leo XIV's Encyclical *Magnifica Humanitas*

### 6A. What is an encyclical, and why does this one matter?

Back in section 1A, we said that the Catholic Church encourages technology "when it respects ethical principles and upholds the dignity of human life," and that "moral discernment is crucial to guide technology to serve humanity ethically." In 2026, the Church put that principle into its highest form of teaching.

On **May 15, 2026**, Pope Leo XIV signed *Magnifica Humanitas* ("The Magnificent Greatness of Humanity") — his **first encyclical**, and the **first encyclical in the Church's history devoted entirely to artificial intelligence**. It was formally presented to the world on **May 25, 2026**. An encyclical is one of the most authoritative forms of papal teaching, so the fact that a Pope chose AI as the subject of his very first one tells you how urgent the Church considers this question to be.

Pope Leo XIV is not writing as a computer scientist. He is writing as a pastor asking: *What does AI mean for the dignity of the human person?*

### 6B. What does the Pope say AI actually *is*?

This is the heart of the question we started this class with — "What is AI to you?" Here is how Pope Leo XIV answers it, in his own words:

> "So-called artificial intelligences do not undergo experiences, do not possess a body, do not feel joy or pain," and cannot truly understand love, work, or friendship.

In other words, the Pope draws a sharp line between what a machine can *do* (process information, recognize patterns, generate text or images) and what a human being *is* (an embodied person who experiences the world, forms relationships, and loves). AI can imitate the outputs of intelligence — but it does not have a body, does not suffer, does not rejoice, and does not love.

He also addresses technology in general — a helpful reminder that AI is not something totally separate from the rest of human history:

> "Technology should not be considered, in itself, as a force antagonistic to humanity. On the contrary, it has formed part of our history since the beginning as 'a profoundly human reality, linked to the autonomy and freedom of man.'"

But he is careful to note that tools are never morally "neutral" the way a hammer might seem to be:

> "Technology is never neutral, because it takes on the characteristics of those who devise, finance, regulate and use it."

And on what makes the human person irreplaceable, no matter how advanced AI becomes:

> "We must lovingly safeguard the grandeur of humanity bestowed upon us... the splendor of which no machine can ever replace."

**Big idea:** For Pope Leo XIV, AI can simulate *some outputs* of intelligence, but it does not have consciousness, a body, feelings, or the capacity to love — the very things that make a human being a person made in the image of God.

### 6C. Key teachings and warnings from the encyclical

- **A choice between two paths.** The Pope frames our moment as a choice: "between constructing Babel or rebuilding Jerusalem; between a power that claims to dominate the heavens and a people who work together in the presence of God to rebuild the walls of fraternal coexistence." He warns against the "Babel syndrome" — "the idolatry of profit that sacrifices the weak, a uniformity that neutralizes differences, and the pretense that a single language — even a digital one — can translate everything."

- **AI and war.** He teaches that some decisions must never be handed to a machine: "it is not permissible to entrust lethal or otherwise irreversible decisions to artificial systems." He calls for humanity to "disarm" AI — not to reject the technology, but to prevent it from dominating human decisions, especially decisions about life and death.

- **AI and work.** Automation and job loss "must not be evaluated solely in terms of efficiency, but in relation to the dignity of the worker, the right to sufficient remuneration." A worker is not just a productivity number.

- **AI and truth.** An entire chapter of the encyclical is devoted to how AI-driven communication and algorithms shape what we believe is true, warning that disinformation amplified by algorithms can weaken democratic society over time.

- **AI and inequality.** Since a small number of private, often transnational companies control most AI development — with resources that "surpass those of many Governments" — the Pope argues that things like "patents, algorithms, digital platforms, technological infrastructure and data" should serve the common good of everyone, not just be concentrated in a few hands.

- **Human weakness is not a flaw to fix.** Against the idea (sometimes called *transhumanism*) that technology should help humans overcome all their limitations, the Pope teaches: "Building for the common good means accepting the limits and weakness of humanity without considering them an error to be corrected."

## 8. Vocabulary

- **Artificial Intelligence (AI)**
The field of computer science that focuses on creating systems capable of performing tasks that typically require human intelligence, such as problem-solving, understanding language, recognizing patterns, and making decisions.

- **Computer Science**
The broad field of study that includes programming, computing systems, software, and technologies such as artificial intelligence.

- **Deep Learning (DL)**
A subset of machine learning that uses artificial neural networks to model complex patterns.

- **Generative AI**
AI systems that can create new content such as text, images, music, or other media.

- **Machine Learning (ML)**
A subset of AI that uses data and algorithms to learn from experience.

- **Narrow AI**
AI that is designed to perform one specific task or a narrow set of tasks.

- **Neural Networks**
A type of computing model used in deep learning that is inspired by the structure of the human brain.

- **Superintelligence AI**
A hypothetical AI that would be more capable than humans at essentially all intellectual tasks.

- **Supervised Learning**
A type of machine learning that learns from data using labels.

- **Unsupervised Learning**
A type of machine learning in which the system looks for patterns in data without being given labels.



- **Encyclical**
A formal, in-depth teaching letter written by a Pope and addressed to the whole Church (and often, on major moral questions, to "all people of good will"). Encyclicals carry significant weight in Catholic moral and social teaching.

- **Human Dignity**
The Catholic teaching that every human person has inherent, irreplaceable worth because he or she is made in the image and likeness of God — a worth that no technology, including AI, can possess or replace.

- **Magnifica Humanitas**
Latin for "the magnificent [greatness of] humanity." The first encyclical devoted specifically to artificial intelligence, written by Pope Leo XIV, signed on May 15, 2026, and formally presented on May 25, 2026.
